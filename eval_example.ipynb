# Dummy Example
Here we will just random noise to the eval data. In practice, the predictions would be generated by a stock prediction model.

## Create a synthetic set of stock predictions

```python
# normally we wouldn't look at eval data at all in training in tuning. This is a dummy example.
eval_data = collect_eval_data()
eval_data.head()
```

```python
# preds would normally come from a deep forecasting model
preds = eval_data + np.random.normal(size=eval_data.shape)
preds.head()
```

## Call the StockModelEvaluator 
All we have to feed are the predictions and the function will independently access the eval dataset and return metrics

```python
evaluator = StockModelEvaluator()
evaluator.eval(preds)
```

Evaluate StockLSTM

```python
preds_lstm, results_lstm, lstm_model, lstm_scaler = train_eval(
    model_class=StockLSTM,
    kwargs={
        "hidden_dim": 64
    },
    seq=60,
    epochs=100,
    lr=1e-3
)
```
